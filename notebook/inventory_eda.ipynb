```python
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
import os
import warnings

warnings.filterwarnings('ignore')

print('Libraries loaded.')
print(f'Pandas version    : {pd.__version__}')
print(f'Matplotlib version: {plt.matplotlib.__version__}')
```

```python
# Build the path to the cleaned CSV file
notebook_dir = os.getcwd()
project_root = os.path.dirname(notebook_dir)
cleaned_csv  = os.path.join(project_root, 'data', 'cleaned', 'inventory_cleaned.csv')

# Load it
df = pd.read_csv(cleaned_csv)

# Quick confirmation
print(f'Loaded: {df.shape[0]} rows x {df.shape[1]} columns')
print(f'Columns: {list(df.columns)}')
```

```python
# Set the visual style for all Matplotlib charts in this notebook
plt.style.use('seaborn-v0_8-whitegrid')

# Set default figure size — width x height in inches
plt.rcParams['figure.figsize'] = (12, 6)

# Set default font size for all text in charts
plt.rcParams['font.size'] = 12

# Set a professional color palette we will use consistently
COLORS = {
    'primary'   : '#2563EB',   # strong blue — main bars and lines
    'secondary' : '#16A34A',   # green — positive metrics
    'warning'   : '#D97706',   # amber — low stock alerts
    'danger'    : '#DC2626',   # red — out of stock
    'neutral'   : '#6B7280',   # gray — background elements
}

# A list of distinct colors for multi-category charts
CATEGORY_COLORS = [
    '#2563EB', '#16A34A', '#D97706', '#DC2626', '#7C3AED',
    '#0891B2', '#DB2777', '#65A30D', '#EA580C', '#0D9488',
    '#9333EA', '#B45309', '#475569'
]

print('Chart style configured.')
print(f'Default figure size : {plt.rcParams["figure.figsize"]}')
print(f'Default font size   : {plt.rcParams["font.size"]}')
print(f'Color palette       : {len(CATEGORY_COLORS)} distinct colors defined')
```

```python
# Count how many products are in each category
# value_counts() counts occurrences of each unique value and sorts by count descending
category_counts = df['Category'].value_counts()

print('Products per category:')
print(category_counts)
```

```python
# Create the horizontal bar chart
fig, ax = plt.subplots(figsize=(12, 7))

# Plot horizontal bars
# category_counts.values = the numbers (bar lengths)
# category_counts.index  = the category names (y-axis labels)
bars = ax.barh(
    category_counts.index,
    category_counts.values,
    color=CATEGORY_COLORS[:len(category_counts)],
    edgecolor='white',
    height=0.6
)

# Add the exact number at the end of each bar
# This makes the chart readable without needing to look at the axis
for bar, value in zip(bars, category_counts.values):
    ax.text(
        bar.get_width() + 0.3,    # x position: just past the end of the bar
        bar.get_y() + bar.get_height() / 2,  # y position: middle of the bar
        str(value),               # the text to display
        va='center',              # vertical alignment: center
        fontsize=11,
        fontweight='bold'
    )

# Labels and title
ax.set_xlabel('Number of Products', fontsize=12)
ax.set_title('Number of Products per Category\nNate Data — Electrical Shop Inventory',
             fontsize=14, fontweight='bold', pad=15)

# Remove the top and right border lines — cleaner look
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Add a little padding on the right so the numbers don't get cut off
ax.set_xlim(0, category_counts.values.max() + 5)

plt.tight_layout()
plt.show()
```

```python
# Group by category and sum the total stock value for each
# This tells us: how much total ETB is invested in each category
value_by_category = (
    df.groupby('Category')['Total_Stock_Value_ETB']
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

# Rename columns for clarity
value_by_category.columns = ['Category', 'Total_Value_ETB']

# Add a percentage column — what share of total inventory value does each category hold
total = value_by_category['Total_Value_ETB'].sum()
value_by_category['Percentage'] = (
    (value_by_category['Total_Value_ETB'] / total * 100).round(1)
)

print('Inventory value by category:')
print()
print(value_by_category.to_string(index=False))
```

```python
fig, ax = plt.subplots(figsize=(12, 7))

bars = ax.bar(
    value_by_category['Category'],
    value_by_category['Total_Value_ETB'],
    color=CATEGORY_COLORS[:len(value_by_category)],
    edgecolor='white',
    width=0.6
)

# Add value labels on top of each bar
for bar, value, pct in zip(bars,
                            value_by_category['Total_Value_ETB'],
                            value_by_category['Percentage']):
    ax.text(
        bar.get_x() + bar.get_width() / 2,  # center of the bar horizontally
        bar.get_height() + 1500,             # just above the top of the bar
        f'{value:,.0f}\n({pct}%)',           # value in ETB + percentage
        ha='center',                          # horizontal alignment: center
        va='bottom',                          # vertical alignment: bottom
        fontsize=9,
        fontweight='bold'
    )

# Rotate x-axis labels so they don't overlap
plt.xticks(rotation=35, ha='right', fontsize=10)

ax.set_ylabel('Total Stock Value (ETB)', fontsize=12)
ax.set_title('Total Inventory Value by Category\nNate Data — Electrical Shop Inventory',
             fontsize=14, fontweight='bold', pad=15)

# Format y-axis numbers with comma separators
ax.yaxis.set_major_formatter(
    plt.FuncFormatter(lambda x, _: f'{x:,.0f}')
)

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Add a horizontal line showing the average category value
average_value = value_by_category['Total_Value_ETB'].mean()
ax.axhline(
    y=average_value,
    color=COLORS['danger'],
    linestyle='--',
    linewidth=1.5,
    label=f'Average: {average_value:,.0f} ETB'
)
ax.legend(fontsize=10)

plt.tight_layout()
plt.show()
```

```python
# Summary statistics for price column
print('Purchase Price Statistics:')
print()
print(f'  Lowest price    : {df["Purchase_Price_ETB"].min():,} ETB')
print(f'  Highest price   : {df["Purchase_Price_ETB"].max():,} ETB')
print(f'  Average price   : {df["Purchase_Price_ETB"].mean():,.0f} ETB')
print(f'  Median price    : {df["Purchase_Price_ETB"].median():,.0f} ETB')
print(f'  Std deviation   : {df["Purchase_Price_ETB"].std():,.0f} ETB')
print()

# Price buckets — how many products fall in each range
bins   = [0, 200, 500, 1000, 2000, 5000]
labels = ['0–200', '201–500', '501–1000', '1001–2000', '2001–5000']

price_buckets = pd.cut(
    df['Purchase_Price_ETB'],
    bins=bins,
    labels=labels
)

print('Products per price range:')
print()
print(price_buckets.value_counts().sort_index())
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# --- Left chart: full histogram ---
axes[0].hist(
    df['Purchase_Price_ETB'],
    bins=30,
    color=COLORS['primary'],
    edgecolor='white',
    linewidth=0.8
)

# Add vertical lines for mean and median
axes[0].axvline(
    df['Purchase_Price_ETB'].mean(),
    color=COLORS['danger'],
    linestyle='--',
    linewidth=2,
    label=f'Mean: {df["Purchase_Price_ETB"].mean():,.0f} ETB'
)
axes[0].axvline(
    df['Purchase_Price_ETB'].median(),
    color=COLORS['secondary'],
    linestyle='--',
    linewidth=2,
    label=f'Median: {df["Purchase_Price_ETB"].median():,.0f} ETB'
)

axes[0].set_xlabel('Purchase Price (ETB)', fontsize=11)
axes[0].set_ylabel('Number of Products', fontsize=11)
axes[0].set_title('Price Distribution — All Products', fontsize=12, fontweight='bold')
axes[0].legend(fontsize=10)
axes[0].spines['top'].set_visible(False)
axes[0].spines['right'].set_visible(False)

# --- Right chart: price bucket bar chart ---
bucket_counts = price_buckets.value_counts().sort_index()

axes[1].bar(
    bucket_counts.index,
    bucket_counts.values,
    color=CATEGORY_COLORS[:len(bucket_counts)],
    edgecolor='white',
    width=0.6
)

# Add count labels on top of each bar
for i, (label, value) in enumerate(zip(bucket_counts.index, bucket_counts.values)):
    pct = value / len(df) * 100
    axes[1].text(
        i,
        value + 0.5,
        f'{value}\n({pct:.0f}%)',
        ha='center',
        fontsize=10,
        fontweight='bold'
    )

axes[1].set_xlabel('Price Range (ETB)', fontsize=11)
axes[1].set_ylabel('Number of Products', fontsize=11)
axes[1].set_title('Products per Price Range', fontsize=12, fontweight='bold')
axes[1].spines['top'].set_visible(False)
axes[1].spines['right'].set_visible(False)

plt.suptitle('Nate Data — Purchase Price Analysis',
             fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
# Define stock level thresholds
LOW_STOCK_THRESHOLD = 5   # 5 or fewer units = low stock

# Categorize every product by stock level
# We create a new column just for this analysis (not saved to the dataset)
df['Stock_Status'] = 'Healthy'
df.loc[df['Quantity'] <= LOW_STOCK_THRESHOLD, 'Stock_Status'] = 'Low Stock'
df.loc[df['Quantity'] == 0, 'Stock_Status'] = 'Out of Stock'

# Summary counts
out_of_stock = df[df['Stock_Status'] == 'Out of Stock']
low_stock    = df[df['Stock_Status'] == 'Low Stock']
healthy      = df[df['Stock_Status'] == 'Healthy']

print('Stock Level Summary:')
print()
print(f'  Out of Stock (Qty = 0)     : {len(out_of_stock)} products')
print(f'  Low Stock   (Qty 1–5)      : {len(low_stock)} products')
print(f'  Healthy     (Qty > 5)      : {len(healthy)} products')
print()

# Show out of stock products in detail
print('Out of Stock Products:')
print()
print(out_of_stock[['Product_ID', 'Product_Name', 'Category',
                     'Purchase_Price_ETB']].to_string(index=False))
print()

# Show low stock products in detail
print('Low Stock Products (need attention soon):')
print()
print(low_stock[['Product_ID', 'Product_Name', 'Category',
                  'Quantity', 'Purchase_Price_ETB']]
      .sort_values('Quantity')
      .to_string(index=False))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# --- Left chart: stock status overview (donut chart) ---
status_counts = df['Stock_Status'].value_counts()

status_colors = {
    'Healthy'     : COLORS['secondary'],
    'Low Stock'   : COLORS['warning'],
    'Out of Stock': COLORS['danger']
}

pie_colors = [status_colors[s] for s in status_counts.index.tolist()]

wedges, texts, autotexts = axes[0].pie(
    status_counts.values.tolist(),
    labels=status_counts.index.tolist(),
    colors=pie_colors,
    autopct='%1.1f%%',
    startangle=90,
    wedgeprops={'width': 0.6}   # width < 1 makes it a donut chart
)

# Make percentage text bold
for autotext in autotexts:
    autotext.set_fontweight('bold')
    autotext.set_fontsize(11)

axes[0].set_title('Stock Status Overview', fontsize=12, fontweight='bold')

# Add center text showing total products
axes[0].text(0, 0, f'{len(df)}\nProducts',
             ha='center', va='center',
             fontsize=13, fontweight='bold')

# --- Right chart: low stock products bar chart ---
low_stock_sorted = low_stock.sort_values('Quantity')

bar_colors = [COLORS['danger'] if q == 0 else COLORS['warning']
              for q in low_stock_sorted['Quantity'].tolist()]

bars = axes[1].barh(
    low_stock_sorted['Product_Name'].tolist(),
    low_stock_sorted['Quantity'].tolist(),
    color=bar_colors,
    edgecolor='white',
    height=0.6
)

# Add quantity labels
for bar, value in zip(bars, low_stock_sorted['Quantity'].tolist()):
    axes[1].text(
        bar.get_width() + 0.05,
        bar.get_y() + bar.get_height() / 2,
        str(value),
        va='center',
        fontsize=10,
        fontweight='bold'
    )

axes[1].set_xlabel('Quantity in Stock', fontsize=11)
axes[1].set_title('Low Stock Products (Qty ≤ 5)',
                   fontsize=12, fontweight='bold')
axes[1].set_xlim(0, LOW_STOCK_THRESHOLD + 2)
axes[1].spines['top'].set_visible(False)
axes[1].spines['right'].set_visible(False)
axes[1].axvline(x=0, color=COLORS['danger'],
                linewidth=1.5, linestyle='--', alpha=0.5)

plt.suptitle('Nate Data — Stock Level Analysis',
             fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
# Sum total quantity per category
qty_by_category = (
    df.groupby('Category')['Quantity']
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

qty_by_category.columns = ['Category', 'Total_Quantity']

# Add percentage of total units
total_units = qty_by_category['Total_Quantity'].sum()
qty_by_category['Percentage'] = (
    (qty_by_category['Total_Quantity'] / total_units * 100).round(1)
)

print(f'Total units across all categories: {total_units:,}')
print()
print(qty_by_category.to_string(index=False))
```

```python
fig, ax = plt.subplots(figsize=(12, 7))

bars = ax.bar(
    qty_by_category['Category'].tolist(),
    qty_by_category['Total_Quantity'].tolist(),
    color=CATEGORY_COLORS[:len(qty_by_category)],
    edgecolor='white',
    width=0.6
)

# Add labels on top of each bar showing quantity and percentage
for bar, qty, pct in zip(bars,
                          qty_by_category['Total_Quantity'].tolist(),
                          qty_by_category['Percentage'].tolist()):
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + 5,
        f'{qty:,}\n({pct}%)',
        ha='center',
        va='bottom',
        fontsize=9,
        fontweight='bold'
    )

plt.xticks(rotation=35, ha='right', fontsize=10)

ax.set_ylabel('Total Units in Stock', fontsize=12)
ax.set_title('Total Quantity in Stock by Category\nNate Data — Electrical Shop Inventory',
             fontsize=14, fontweight='bold', pad=15)

# Add average line
avg_qty = qty_by_category['Total_Quantity'].mean()
ax.axhline(
    y=avg_qty,
    color=COLORS['danger'],
    linestyle='--',
    linewidth=1.5,
    label=f'Average: {avg_qty:,.0f} units'
)

ax.legend(fontsize=10)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
# Check if there are any extreme outliers
print('Price and Quantity Distribution:')
print()
print(f'  Price   — Min: {df["Purchase_Price_ETB"].min():,} ETB, '
      f'Max: {df["Purchase_Price_ETB"].max():,} ETB')
print(f'  Quantity — Min: {df["Quantity"].min()}, '
      f'Max: {df["Quantity"].max()}')
print()

# Products with both high price and high quantity (unusual)
expensive_and_high_qty = df[
    (df['Purchase_Price_ETB'] > 1000) & (df['Quantity'] > 20)
]

if not expensive_and_high_qty.empty:
    print('Products with high price AND high quantity (unusual pattern):')
    print()
    print(expensive_and_high_qty[['Product_Name', 'Category',
                                   'Purchase_Price_ETB', 'Quantity',
                                   'Total_Stock_Value_ETB']].to_string(index=False))
else:
    print('No products with both high price (>1000 ETB) and high quantity (>20).')
```

```python
fig, ax = plt.subplots(figsize=(12, 7))

# Create scatter plot with category colors
# We map each category to a color
category_list = df['Category'].unique().tolist()
color_map = {cat: CATEGORY_COLORS[i] for i, cat in enumerate(category_list)}

for category in category_list:
    cat_data = df[df['Category'] == category]
    ax.scatter(
        cat_data['Purchase_Price_ETB'].tolist(),
        cat_data['Quantity'].tolist(),
        c=color_map[category],
        label=category,
        alpha=0.6,
        s=80,
        edgecolors='white',
        linewidth=0.5
    )

ax.set_xlabel('Purchase Price (ETB)', fontsize=12)
ax.set_ylabel('Quantity in Stock', fontsize=12)
ax.set_title('Price vs Quantity — Inventory Pattern Analysis\n'
             'Nate Data — Electrical Shop',
             fontsize=14, fontweight='bold', pad=15)

# Format x-axis with commas
ax.xaxis.set_major_formatter(
    plt.FuncFormatter(lambda x, _: f'{x:,.0f}')
)

# Add grid for easier reading
ax.grid(True, alpha=0.2, linestyle='--')

# Legend outside the plot area so it doesn't cover data
ax.legend(bbox_to_anchor=(1.05, 1), loc='upper left',
          fontsize=9, frameon=False)

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
```
