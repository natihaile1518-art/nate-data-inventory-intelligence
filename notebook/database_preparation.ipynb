```python
import pandas as pd
import os

print("Libraries loaded.")
print(f"Pandas version: {pd.__version__}")
```

```python
# Build the path to the cleaned inventory file
notebook_dir = os.getcwd()
project_root = os.path.dirname(notebook_dir)
csv_path     = os.path.join(project_root, 'data', 'cleaned', 'inventory_cleaned.csv')

# Load it
df = pd.read_csv(csv_path)

# Immediate confirmation
print(f"File loaded: {csv_path}")
print(f"Shape: {df.shape[0]} rows x {df.shape[1]} columns")
```

```python
# Look at the first 5 rows
df.head()
```

```python
# Inspect column names and data types
print("Column names and data types:")
print()
for col, dtype in df.dtypes.items():
    print(f"  {col:<28} {str(dtype)}")
```

```python
# Look at unique values for categorical columns
# This helps us understand what goes into separate tables

print("=== Product_ID ===")
print(f"  Total values : {df['Product_ID'].count()}")
print(f"  Unique values: {df['Product_ID'].nunique()}")
print(f"  Sample       : {df['Product_ID'].head(3).tolist()}")
print()

print("=== Category ===")
print(f"  Unique categories: {df['Category'].nunique()}")
print(f"  Values:")
for cat in sorted(df['Category'].unique()):
    count = (df['Category'] == cat).sum()
    print(f"    {cat:<25} {count} products")
print()

print("=== Location ===")
print(f"  Unique values: {df['Location'].unique().tolist()}")
print()

print("=== Specification ===")
print(f"  Total values  : {df['Specification'].count()}")
print(f"  N/A count     : {(df['Specification'] == 'N/A').sum()}")
print(f"  Unique specs  : {df['Specification'].nunique()}")
print(f"  Sample        : {df['Specification'].head(5).tolist()}")
print()

print("=== Quantity ===")
print(f"  Min   : {df['Quantity'].min()}")
print(f"  Max   : {df['Quantity'].max()}")
print(f"  Zeros : {(df['Quantity'] == 0).sum()} products")
print()

print("=== Purchase_Price_ETB ===")
print(f"  Min   : {df['Purchase_Price_ETB'].min():,} ETB")
print(f"  Max   : {df['Purchase_Price_ETB'].max():,} ETB")
print(f"  Mean  : {df['Purchase_Price_ETB'].mean():,.0f} ETB")
print()

print("=== Count_Date ===")
print(f"  Unique dates: {df['Count_Date'].unique().tolist()}")
print()

print("=== Needs_Review ===")
print(f"  Values: {df['Needs_Review'].value_counts().to_dict()}")
print()

print("=== Total_Stock_Value_ETB ===")
print(f"  Total inventory value: {df['Total_Stock_Value_ETB'].sum():,.0f} ETB")
print(f"  Note: this is a calculated field (Quantity x Price)")
```

```python
# Final readiness check before database work begins
print("=== DATA READINESS SUMMARY ===")
print()
print(f"  Total products to import : {len(df)}")
print(f"  Total categories         : {df['Category'].nunique()}")
print(f"  Products with NULL spec  : {(df['Specification'] == 'N/A').sum()}")
print(f"  Products with zero qty   : {(df['Quantity'] == 0).sum()}")
print(f"  Count date               : {df['Count_Date'].unique()[0]}")
print(f"  Location label           : {df['Location'].unique()[0]}")
print()
print("  Column decisions:")
print("  → products table    : Product_ID, Product_Name, Specification")
print("  → categories table  : Category (13 unique values)")
print("  → locations table   : Location (maps to Shop 1)")
print("  → inventory_snapshots: Quantity, Count_Date")
print("  → NOT imported      : Needs_Review, Total_Stock_Value_ETB")
print("  → purchase_items    : Purchase_Price_ETB (starting reference)")
print()
print("Data is ready for MySQL schema design.")
```
